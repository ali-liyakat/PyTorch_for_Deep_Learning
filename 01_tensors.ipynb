**Creating a Tensor**

```python
# Using empty

x = torch.empty(2, 3)
print(x)
```

```python
# check type

print(type(x))
```

```python
# using zeroes

x = torch.zeros(2, 3)
print(x)
```

```python
# using ones

x = torch.ones(2, 3)
print(x)
```

```python
# using rand

x = torch.rand(2, 3)
print(x)
```

```python
# use of seed

torch.manual_seed(100)
x = torch.rand(2, 3)
print(x)
```

```python
# using tensor

x = torch.tensor([1, 2, 3])
print(x)
```

```python
# using arange

x = torch.arange(0, 10, 2)
print(x)
```

```python
# using linspace

x = torch.linspace(0, 10, 5)
print(x)
```

```python
# using eye

x = torch.eye(3)
print(x)
```

```python
# using full

x = torch.full((2, 3), 2)
print(x)
```

4. Matrix Operations

```python
f = torch.randint(size=(2, 3), low=0, high=10, dtype=torch.float16)
g = torch.randint(size=(3, 2), low=0, high=10, dtype=torch.float16)
```

```python
# Matrix Multiplication
torch.matmul(f, g)
```

```python
# dot product
torch.dot(torch.tensor([1, 2, 3]), torch.tensor([4, 5, 6]))
```

```python
# transpose
torch.transpose(f, 0, 1)
```

```python
# Determinant
torch.det(torch.eye(3))
```

```python
# inverse
torch.inverse(torch.eye(3))
```
